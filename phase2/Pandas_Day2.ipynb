```python
import pandas as pd
import numpy as np


employees = {
    "name":[
        " Ronak ",
        "Amit",
        "Neha",
        "Ronak",
        "Raj"
    ],

    "salary":[
        50000,
        np.nan,
        30000,
        50000,
        90000
    ],

    "dept":[
        "sql",
        "PYTHON",
        "Sql",
        "sql",
        None
    ]
}


df = pd.DataFrame(employees)


print(df)
```

```python
## Count Missing values 
## Exercise 1 
df.isnull().sum()
```

* Difference between dropna and fillna?

    dropna removes missing records.

    fillna replaces missing values with default/calculated values.

```python
## Remove null data from dataframe
clean_data = df.dropna()
print(clean_data)
```

```python
update = df
update["salary"] = (update["salary"].fillna(0))
```

```python
## Exercise 2
update["dept"] = (update["dept"].fillna("Unknown"))
print(update)
```

```python
## Exercise 3
df["name"] = df["name"].str.strip()
df["dept"] = df["dept"].str.strip()

df = df.drop_duplicates(subset=['name','salary','dept'])

print(df)
```

```python
import pandas as pd
import numpy as np


employees={

"name":[
" Ronak ",
"Amit",
"Neha",
"Ronak",
"Raj"
],

"salary":[
50000,
np.nan,
30000,
50000,
90000
],

"dept":[
"sql",
"PYTHON",
"Sql",
"sql",
None
]

}


df=pd.DataFrame(employees)
print(df)
```

```python
df["name"] = df["name"].str.strip()
df["dept"] = df["dept"].str.strip()
df["dept"] = df["dept"].str.upper()

df["salary"] = (df["salary"].fillna(df["salary"].mean()))
df["dept"] = (df["dept"].fillna("UNKNOWN"))

df = df.drop_duplicates()

df["status"] = "CLEANED"
print(df)
```
