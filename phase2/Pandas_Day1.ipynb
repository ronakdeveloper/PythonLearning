```python
## Exercise 3
employees = {
"name":["Ronak","Amit","Neha","Raj"],

"salary":[50000,80000,30000,90000],

"dept":["SQL","Python","SQL","Python"]
}

result = pd.DataFrame(employees)

output = result[
    (result["salary"]>70000)
    |
    (result["dept"]=="Python")
]

print(output)
```

```python
result["annual_salary"] = (
    result["salary"] * 12
)

print(result)
```

```python
result.rename(
    columns={
        "name":"employee_name"
    },
    inplace=True
)

print(result)
```

```python
result.sort_values(
    "salary",
    ascending=False
)
```

```python
import numpy as np

employees = {
"name":["Ronak","Amit","Neha","Raj"],

"salary":[50000,80000,30000,90000],

"dept":["SQL","Python","SQL","Python"]
}

emp = pd.DataFrame(employees)

emp["annual_salary"] = (emp["salary"] * 12)

emp["grade"] = np.where(emp["salary"]>=70000,"Senior","Junior")

Senior_emp = emp[emp["grade"]=="Senior"]
print(Senior_emp)

emp.sort_values(
    "salary",
    ascending=False
)
```
