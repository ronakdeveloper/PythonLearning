```python
import pandas as pd
import numpy as np

def extract(path):
    df = pd.read_csv(path)

    return df

def transform(df):
    df["salary"] = (df["salary"].fillna(0))
    df["annual_salary"] = (df["salary"] * 12)
    df["grade"] = np.where(df["salary"] >= 70000, "Senior","Junior")

    return df

def load(df):
    valid_records = df[df["salary"] > 0]
    invalid_records = df[df["salary"] <= 0]

    valid_records.to_csv("processed.csv",index=False)
    invalid_records.to_csv("rejected.csv",index=False)


def main(Path):
    data = extract(Path)

    data = transform(data)

    load(data)
```

```python
main("employees.csv")
```
