```python
## Aggregations

salary=np.array(
[30000,50000,80000]
)

print(np.max(salary))

print(np.min(salary))

print(np.mean(salary))

print(np.sum(salary))
```

```python
## Where 
salary=np.array(
[30000,80000,40000]
)

result = np.where(salary > 50000,"High","Low")
print(result)
```
