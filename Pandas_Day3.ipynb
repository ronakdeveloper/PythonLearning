```python
import pandas as pd


employees = {
"name":["Ronak","Amit","Neha","Raj","Jay"],

"dept":[
"SQL",
"Python",
"SQL",
"Python",
"SQL"
],

"salary":[
50000,
80000,
30000,
90000,
60000
]
}


df = pd.DataFrame(employees)

print(df)
```

```python
result = (
    df.groupby("dept")["salary"].sum()
)

print(result)
```

```python
## Exercise 1 
avg_salary = (
    df.groupby("dept")["salary"].mean()
)

print(avg_salary)
```

```python
result = (
    df.groupby("dept")
    .agg(
        total_salary = ("salary","sum"),
        avg_salary = ("salary","mean"),
        emp_count = ("name","count")
    )
)

print(result)
```

```python
## Exercise 2 
result = (
    df.groupby("dept").
    agg(
        max_salary = ("salary","max"),
        min_salary = ("salary","min"),
        emp_count = ("name","count")
    )
)

print(result)
```

```python
## Group By Multiple columns 
employees={

"name":["Ronak","Amit","Neha","Raj"],

"dept":["SQL","SQL","Python","Python"],

"city":["Ahmedabad","Surat","Ahmedabad","Surat"],

"salary":[50000,70000,80000,90000]

}


df=pd.DataFrame(employees)


result=(
df.groupby(
[
"dept",
"city"
]
)["salary"]
.sum()
)


print(result)
```

```python
## Join
employees = pd.DataFrame(
{
"emp_id":[1,2,3],

"name":["Ronak","Amit","Neha"],

"dept_id":[10,20,30]
}
)

departments = pd.DataFrame(
{
"dept_id":[10,20],

"dept_name":["SQL","Python"]
}
)

result = pd.merge(employees,departments, on="dept_id",how="inner")
print(result)
```

```python
## Left Join
result=pd.merge(
employees,
departments,
on="dept_id",
how="left"
)

print(result)
```

```python
## Concat
jan = pd.DataFrame(
{
"sale":[100,200]
}
)

feb = pd.DataFrame(
{
"sale":[300,400]
}
)

final = pd.concat([jan,feb])

print(final)
```

```python
orders=pd.DataFrame(
{
"order_id":[1,2,3,4],

"customer_id":[101,102,101,103],

"amount":[
500,
1000,
700,
2000
]
}
)

customers=pd.DataFrame(
{
"customer_id":[101,102],

"name":[
"Ronak",
"Amit"
]
}
)
```

```python
## Total amount customer wise
total_amount = orders.groupby(
    ["customer_id"]
)["amount"].sum()

print(total_amount)
```

```python
## Avg order amount 
avg_order_amount = orders["amount"].mean()

print(avg_order_amount)
```

```python
## Join customer name with orders 
result = pd.merge(orders,customers,on="customer_id",how="inner")

print(result)
```

```python
## Find customer without matching names
## Join customer name with orders 
result = pd.merge(orders,customers,on="customer_id",how="left")

result["name"] = (result["name"].fillna("UNKNOWN"))
final = result[result["name"] == "UNKNOWN"]
print(final)
```
