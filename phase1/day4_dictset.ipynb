```python
employee = {
    "name" : "Ronak",
    "salary" : "50000",
    "department" : "IT"
}

print(employee)
print(employee["name"])
print(employee["salary"])
```

```python
## Add New key
employee["city"] = "Ahmedabad"
print(employee)
```

```python
## Update existing value
employee["salary"] = 60000
print(employee)
```

```python
## Remove Key 
employee.pop("city")
print(employee)
```

```python
## Print Keys and values
print(employee.keys())
print(employee.values())
print(employee.items())
```

```python
## Loop through dic
for key in employee:
    print(key)
```

```python
for value in employee.values():
    print(value)
```

```python
for key,value in employee.items():
    print(key, value)
```

```python
## Nested dict
employee = {
    "name": "Ronak",
    "salary": 50000,
    "address": {
        "city": "Ahmedabad",
        "state": "Gujarat"
    }
}

print(employee)
print(employee["address"]["city"])
```

```python
## What is set  : A set stores unique values.
skills = {"SQL", "Python", "SQL", "Spark"}

print(skills)
```

```python
skills.add("AirFlow")
print(skills)
```

```python
skills.remove("SQL")
print(skills)
```

```python
## Membership check
print("Python" in skills)
```

```python
## Find Unique Employee 
employees = [
    "Ronak",
    "Amit",
    "Ronak",
    "Neha",
    "Amit"
]
print(employees)

unique_employee = set(employees)
print(unique_employee)
```

```python
employees = [
    {"name": "Ronak", "salary": 50000, "dept": "SQL"},
    {"name": "Amit", "salary": 70000, "dept": "Python"},
    {"name": "Neha", "salary": 40000, "dept": "SQL"},
    {"name": "Raj", "salary": 90000, "dept": "Python"}
]

for employee in employees:
    if employee["dept"] == "Python":
        print(employee["name"])
```

```python
employees = [
    {"name": "Ronak", "salary": 50000},
    {"name": "Amit", "salary": 70000},
    {"name": "Neha", "salary": 40000},
    {"name": "Raj", "salary": 90000}
]

for emp in employees:
    if emp["salary"] >= 70000:
        print(emp["name"])

total = 0

for i in employees:
    total += i["salary"]

print(total)

for emp in employees:
    emp["salary"] = emp["salary"] * 1.10

print(employees)
```
