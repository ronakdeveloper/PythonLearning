```python
## Exercise 3
 
with open("employees.csv","r") as file:
    result = csv.DictReader(file)
    result_update = list(map(lambda x:{"name":x["name"],"salary":int(x["salary"]),"dept":x["dept"]},result))
    for emp in result_update:   
        if emp["salary"] >= 50000:
            print(emp["name"])
    
```

```python
## Write csv
import csv

employees=[
{"name":"Ronak","salary":50000},
{"name":"Amit","salary":80000}
]

with open("employees_out.csv","w",newline="") as file:
    columns = ["name","salary"]

    writer = csv.DictWriter(file, fieldnames=columns)

    writer.writeheader()
    writer.writerows(employees)

with open("employees_out.csv","r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row)
```

```python
## Exercise 5
import json

with open("emp.json","r") as file:
    result = json.load(file)
    result = list(map(lambda x:{
        "name":x["name"],
        "salary":x["salary"]*1.20
    },result))

with open("updated_emp.json","w") as file:
    json.dump(result, file, indent=4)
```

```python
## Employee ETL Pipeline
import csv
with open("employees.csv","r") as file:
    reader = csv.DictReader(file)
    result = list(filter(lambda x:x["salary"]>="50000",reader))
    result = list(map(lambda x:{"name":x["name"],
    "salary":int(x["salary"]),
    "status":"processed"},result))

with open("processed_employee.csv","w",newline="") as file:
    columns = ["name","salary","status"]

    writer = csv.DictWriter(file,fieldnames=columns)
    writer.writeheader()
    writer.writerows(result)
```
