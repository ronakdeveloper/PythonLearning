```python
## break function 
number = [10,20,30,40]

for num in number:
    if num == 30:
        break

    print(num)
```

```python
## Continue 
for num in number:
    if num == 30:
        continue

    print(num)
```

```python
## Ignore bad records 
records = ["Ronak","","Makwana"]

for i in records:
    if i == "":
        continue
    print(i)
```

```python
## enumerate()
skills = ["SQL","Python","Spark"]

count = 0

for i in skills:
    print(count, i)
    count+=1

for index,i in enumerate(skills):
    print(index, i)
```

```python
names = ["Ronak","Amit"]
salary = [50000,70000]

for x,y in zip(names,salary):
    print(x,y)
```

```python
## Ex5
employees=[
{"name":"Ronak","salary":50000},
{"name":"Amit","salary":80000},
{"name":"Neha","salary":30000}
]

high_salary = []
low_salary = []

for i in employees:
    if i["salary"] >= 50000:
        high_salary.append(i)
    else:
        low_salary.append(i)

print(high_salary)
print(low_salary)
```
