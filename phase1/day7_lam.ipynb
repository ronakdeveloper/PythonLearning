```python
salaries = [10000,20000,30000,40000,50000]

result = []

for salary in salaries:
    if salary > 30000:
        result.append(salary)

print(result)
```

```python
result = [
    salary for salary in salaries if salary > 30000
]

print(result)
```

```python
## Lambda function
def square(x):
    return x*x

print(square(5))

square = lambda x:x*x
print(square(5))
```

```python
## Sorted 
employees=[
{"name":"Ronak","salary":50000},
{"name":"Amit","salary":80000},
{"name":"Neha","salary":30000}
]

result = sorted(employees, key= lambda emp:emp["salary"])

print(result)
```

```python
result = sorted(employees, key= lambda emp:emp["salary"],reverse=True)
print(result)
```
