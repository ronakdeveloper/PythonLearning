```python
## Exercise 2
employees=[
{"name":"Ronak","salary":"50000"},
{"name":"Amit","salary":"abc"},
{"name":"Neha"}
]
valid_records = []
bad_records = []
for employee in employees:
    try:
        valid_records.append({"name":employee["name"],"salary":int(employee["salary"])})
    except ValueError:
        bad_records.append({"name":employee["name"],"error":"ValueError"})
    except KeyError:
        bad_records.append({"name":employee["name"],"error":"KeyError"})
    except Exception as error:
        bad_records.append({"name":employee["name"],"error":error})

print(valid_records)
print(bad_records)
```

```python
def extract(filename):
    import csv
    with open(filename, "r") as file:
        reader = csv.DictReader(file)
        result = []
        for row in reader:
            result.append(row)

    return result

def transform(records):
    valid_records = []
    bad_records = []

    for record in records:
        try:
            if record["salary"] == "":
                raise KeyError()

            salary = int(record["salary"])
            valid_records.append({
                "name":record["name"],
                "salary":salary
            })
        except ValueError:
            bad_records.append({
                "name":record["name"],
                "error":"Invalid salary"
            })
        except KeyError:
            bad_records.append({
                "name":record["name"],
                "error":"Missing salary"
            })
    return valid_records,bad_records
        

employee = extract("employees.csv")

transform(employee)
```
