```python
## Modify List
skills[1] = 'Azure'
print(skills)
```

```python
## Add iteam in list
skills.append("Database")
skills.insert(1,"MySQL")
print(skills)
```

```python
## Remove from list
skills.remove("MySQL")
skills.pop(1)
print(skills)
```

```python
## reverse 
num.reverse()
print(num)
```
