```python
## Ex5
employees = [
    {"name":"Ronak","salary":50000},
    {"name":"Amit","salary":80000},
    {"name":"Neha","salary":30000}
]

def apply_increment(employee):
    employee["salary"] = round(employee["salary"]* 1.10)


for employee in employees:
    apply_increment(employee)

print(employees)
```

```python
def calculate_final_salary(monthly_salary,tax=0.10):
    annual_salary = monthly_salary * 12
    annual_salary -= annual_salary * tax
    return annual_salary

print(calculate_final_salary(50000))
    
```
